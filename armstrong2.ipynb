```python
#another armstrong program to find armstrong numbers in a range/interval
lower = 100
upper = 999
print("Armstrong numbers between",lower,"and",upper,"are:")
for num in range(lower,upper+1):
    no_of_digits = len(str(num))
    sum = 0
    temp = num
    while temp > 0:
        digit = temp % 10
        sum += digit ** no_of_digits
        temp //= 10
    if num == sum:
        print(num)
```
