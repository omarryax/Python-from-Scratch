```python
num = int(input("enter number: "))
factorial = 1
if num<0:
    print("No factorial possible for Negative numbers")
elif num==0:
    print("Factorial of 0 is 1")
elif num>0:
    for i in range(1,num+1):
        factorial= factorial*i
        print (i,"!", "=", factorial,) # showing all factorials thats why printing within loop
    
    print("Factorial of",num,"is",factorial) # to show specific factorial at the end after updated factorial value
```
