```python
array = [1, 2, 3, 4, 5]
sumofarray= sum(array)
print("The sum of the array is:", sumofarray)
```

```python
def sum_of_array(array):
    total = 0
    for element in array:
        total += element
    return total
total_sum = sum_of_array(array)
print ("The sum of the array is:", total_sum)
```
