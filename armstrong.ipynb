```python
import random 
num = random.randint(1,999)

no_of_digits = len(str(num))
print("Number is:",num)
print ("No of digits:",no_of_digits)
```

```python
#initializing vars
sum_of_powers=0
temp_num = num

while temp_num>0:
    digit = temp_num % 10
    sum_of_powers += digit**no_of_digits
    temp_num //= 10

if sum_of_powers==num:
    print(num,"is an Armstrong number")
else: 
    print(num,"is not an Armstrong number")
```
