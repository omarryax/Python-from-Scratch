Python dictionary is an unordered collection of items
Each item of a dictionary has a key/value pair
Dictionaries are optimized to retrieve values when the key is known

```python
#creating empty dictionary
my_dict = {}
```

```python
my_dict = {"name": "John ", "age": 39}
```

```python
my_dict = {1: "john", 2: "doe", 3: "jane"}
```

```python
my_dict = {"name": "John", 1: [1,2,3,4]}
```

```python
#we can use dict()
my_dict = dict({"name": "John", "age": 39})   
```

```python
#accessing dict elements
dict_var = {'name': 'Umar', 'age': 23}
#get() func 
print(dict_var.get('name'))
#using [] operator
print (dict_var['age'])
```

```python
#modifying elements
#using '=' operator 
my_dict['name']= 'Umar'
print (my_dict)
```

```python
#To add a new key-value pair
my_dict['address']='Islamabad'
print(my_dict)
```

```python
#remove element from dict
#pop() method
my_dict.pop('address')
print(my_dict)
```

```python
dict_var.popitem() # remove the last item from the dictionary
print(dict_var)
```

```python
#clear() method used to clear whole dictionary
dict_var.clear()
print(dict_var)
```
