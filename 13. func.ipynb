Types of Function Arguments:
1. Positional Arg
2. Keyword Arg
3. Default Arg
4. Variable Length Positional Arg(*args)
5. Variable Length Keyword Arg(**kwargs)

```python
#Positional arg
def pos_func(name, age):
    print (name, 'is', age, 'years old')
pos_func('Umar', 23 ) #you have to keep order of arguments same as defined in function
```

```python
#keyword arg/#default arg 
def func2(name= 'Umar', age=23):
    print (name, 'is', age, 'years old')
```

```python
func2() #default values will be used
```

```python
"""Variable length arguments are useful when you want to create functions that
take unlimited number of arguments"""
def var_len(*args):
    print("total args",args)
    print("first arg",args[0]) #first argument only
    print("length",len(args)) #length of args
    for i in range(len(args)): #we can iterate through args
        print("loop",args[i])
var_len(1, 2, 3, 4) #you can pass unlimited number of arguments
```

```python
#var length keyword arguments
"""collects args into a new dictionary, where the argument names are the keys
and their values are the corresponding dictionary values"""
def var_kw(**kwargs):
    print (kwargs)
    print("name:", kwargs['name'])#You can access the value of items by giving the key
    print(type(kwargs))
    for key, value in kwargs.items():
        print (key,value)
var_kw(name= 'Umar', city= "ISB")
```

```python
#return value from statement 
def mul(a,b):
    return a*b, a-b, a+b
x= mul(2,3)
print(x) #prints 6
```

```python
"""When you return multiple values, Python actually packs them in a
single tuple and returns it
You can then use multiple assignment to unpack the parts of the returned
tuple"""
def cal(a,b):
    return a*b, a-b, a+b
mul, sub, add = cal(2,3) #unpacking the tuple
print (mul)
print(sub)
print(add)
```

```python
#assigning function to variables
mul = cal
cal(2,3)
```
