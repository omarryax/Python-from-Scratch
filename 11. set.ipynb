```python
#right way to create set
var_set={}
print (var_set) 
type(var_set)
```

```python
#creating empty set
var_set = set()
print (var_set)
type (var_set)
```

```python
#modifying  elements
var_set.update({1,2,3})
print (var_set)
```

```python
#removing elements
var_set.remove(2)
print(var_set)
```

```python
var_set.discard(3)
print(var_set)
```

The only difference between the two is that the discard() function leaves a
set unchanged if the element is not present in the set

On the other hand, the remove() function will raise an error in such a
condition if element is not present in the set
