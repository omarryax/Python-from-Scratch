```python
n= int(input("Enter a number: "))
total_sum = 0
for i in range(1,n+1):
    total_sum = total_sum + i**3
print("The sum of cubes of first",n,"natural numbers is:",total_sum)
```

```python
#for recursion method
def sumofcubes(n):
    if n == 1:
        return 1
    else:
        total = sum(i**3 for i in range(1,n+1))
        return total

if n <= 0:
    print("Enter a positive integer")
else: 
    result = sumofcubes(n)
    print("The sum of cubes of first",n,"natural numbers is:",result)
```
